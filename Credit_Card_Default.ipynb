```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Data Preprocessing**

```python
credit_card = pd.read_csv('UCI_Credit_Card.csv')
credit_card.head()
```

```python
# Drop ID
credit_card = credit_card.drop('ID', axis = 1)
credit_card.head()
```

```python
# One Hot Encoding on SEX variable
credit_card = pd.get_dummies(credit_card, columns = ['SEX'])
credit_card.head()
```

```python
# One Hot Encoding for EDUCATION variable but let's rename columns too
credit_card = pd.get_dummies(credit_card, columns = ['EDUCATION'])
credit_card = credit_card.drop('EDUCATION_0', axis = 1)
credit_card.head()
```

```python
# One Hot Encoding on MARRIAGE variable
credit_card = pd.get_dummies(credit_card, columns = ['MARRIAGE'])
credit_card.head()
```
